```python
df = sns.load_dataset('titanic')
```

```python
sns.countplot(x='survived', data=df)
plt.show()
# Observation:
# Majority of passengers did not survive the disaster.

```

```python
sns.countplot(x='sex', hue='survived', data=df)
plt.title("Survival Count by Gender")
plt.show()

# Observation:
# Females had a higher survival rate compared to males.
```

```python
sns.countplot(x='pclass', hue='survived', data=df)
plt.title("Survival Count by Passenger Class")
plt.show()
# Observation:
# Passengers in 1st class had a higher survival rate than those in 2nd and 3rd class.
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['age'].dropna(), bins=30, kde=True)
plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
# Observation:
# Most passengers were between 20 and 40 years old.
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
# Observation:
# Survival shows positive correlation with fare and negative correlation with passenger class.
```

```python
# Final Conclusion:
# The exploratory data analysis shows that survival on the Titanic
# was strongly influenced by passenger class, gender, and fare.
# First-class passengers and those who paid higher fares had
# better chances of survival.
```
